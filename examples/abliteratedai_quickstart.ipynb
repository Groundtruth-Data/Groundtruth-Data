# Preview an AbliteratedAI chat request
Run both cells in a browser with a Python notebook runtime. Edit the prompt, inspect the JSON your client would send, then extract an answer from an example response. Both cells run locally in the notebook runtime and require no API key or credits.

```python
import json

prompt = "Give me a descriptive error message for an empty CSV upload."
endpoint = "https://api.abliteratedai.dev/v1/chat/completions"
payload = {
    "model": "abliterated-235b-a22b-fp8",
    "messages": [{"role": "user", "content": prompt}],
    "max_tokens": 128,
    "stream": False,
}
print("POST", endpoint)
print(json.dumps(payload, indent=2))
```

## Read the answer
This hand-written fixture illustrates the response shape. A live response can have different text and usage counts.

```python
fixture = {"choices": [{"message": {"role": "assistant", "content": "Your CSV has no rows. Upload a file with a header and at least one data row."}}]}
print(fixture["choices"][0]["message"]["content"])
```
